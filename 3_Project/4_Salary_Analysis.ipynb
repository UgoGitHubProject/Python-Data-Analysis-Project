### How well do jobs and skills pay for Data Analysts?

#### Methodology
>  ##### Evaluate median salary for top 6 data jobs
>  ##### Find median salary per skill for Data Analysts
>  ##### Visualize for highest paying skills and most demanded skills

```python
# Importing Libraries

import ast
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data CleanUp
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x : ast.literal_eval(x) if pd.notna(x) else x)
```

```python
# Filter for data jobs in the United States only.
# In the filtered dataset, drop the records that have missing values for salary_year_avg

df_US = df[df['job_country'] == 'United States'].dropna(subset = ['salary_year_avg'])
```

```python
# Obtain the top 6 jobs in the dataset

job_titles = df_US['job_title_short'].value_counts().index[:6].tolist()

job_titles
```

```python
df['job_title_short'].unique()
```

```python
# Filter the dataset used, for the top 6 jobs only.
# That is Filter the US job dataset for the top 6 jobs

df_US_top6 = df_US[df_US['job_title_short'].isin(job_titles)]

df_US_top6
```

```python
# Make a box plot of the salary values for the various job titles

sns.boxplot(data = df_US_top6, x = 'salary_year_avg', y = 'job_title_short')

sns.set_theme(style = 'ticks')

plt.title('Salary Distribution for Data Jobs in United States')
plt.ylabel('')
plt.xlabel('Yearly Salary (USD)')
plt.xlim(0, 600000)
ticks_x = plt.FuncFormatter(lambda x, _ : f'${int(x/1000)}k')
plt.gca().xaxis.set_major_formatter(ticks_x)
plt.show()
```

```python
df_US_top6_job_count = df_US_top6['job_title_short'].value_counts(ascending = False)

df_US_top6_job_count
```

```python
job_order = (df_US_top6
            .groupby('job_title_short')['salary_year_avg']
            .median()
            .sort_values(ascending = False)
            .index
)
            

job_order
```

```python
sns.boxplot(data = df_US_top6, x = 'salary_year_avg', y = 'job_title_short', order = job_order)

sns.set_theme(style = 'ticks')

plt.title('Salary Distribution for Data Jobs in United States')
plt.ylabel('')
plt.xlabel('Yearly Salary (USD)')
plt.xlim(0, 600000)
ticks_x = plt.FuncFormatter(lambda x, _ : f'${int(x/1000)}k')
plt.gca().xaxis.set_major_formatter(ticks_x)
plt.show()
```

## Investigate Median Salary v/s Skill for Data Analysts

```python
# Get the data for Data Analysts in the United States

df_DA_US = df[(df['job_country'] == "United States") & (df['job_title_short'] == 'Data Analyst')].copy()

# Drop the records where the salary values are missing

df_DA_US = df_DA_US.dropna(subset = ['salary_year_avg'])
```

```python
# Explode the skills column

df_DA_US = df_DA_US.explode('job_skills')

df_DA_US[['salary_year_avg', 'job_skills']]
```

```python
df_skill_pay = df_DA_US.groupby('job_skills')['salary_year_avg'].agg(['count', 'median'])

df_skill_pay
```

```python
df_DA_top_pay =  df_skill_pay.sort_values(by = 'median', ascending = False)

df_DA_top_pay = df_DA_top_pay.head(10)

df_DA_top_pay
```

```python
df_DA_top_skills = df_skill_pay.sort_values(by = 'count', ascending = False)

df_DA_top_skills = df_DA_top_skills.head(10).sort_values(by = 'median', ascending = False)

df_DA_top_skills
```

```python
fig, ax = plt.subplots(2, 1)

sns.set_theme(style = 'ticks')

# Top 10 Highest Paid Skills for Data Analysts
sns.barplot(data = df_DA_top_pay, x = 'median', y = df_DA_top_pay.index, hue = 'median', ax = ax[0], palette = 'dark:b_r')
ax[0].legend().remove()
ax[0].set_title('Top 10 Highest paid Skills for Data Analysts')
ax[0].set_ylabel('')
ax[0].set_xlabel('')
ax[0].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f'${int(x/1000)}K'))

# Top 10 Most In-Demand Skills for Data Analysts
sns.barplot(data = df_DA_top_skills, x = 'median', y = df_DA_top_skills.index, hue = 'median', ax = ax[1], palette = 'light:b')
ax[1].legend().remove()
ax[1].set_title('Top 10 Most In-Demand Skills for Data Analysts')
ax[1].set_ylabel('')
ax[1].set_xlabel('')
ax[1].set_xlim(ax[0].get_xlim())
ax[1].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f'${int(x/1000)}K'))

plt.tight_layout()
```

```python
fig, ax = plt.subplots(2, 1)

sns.set_theme(style = 'ticks')

# Top 10 Highest Paid Skills for Data Analysts
sns.barplot(data = df_DA_top_pay, x = 'median', y = 'job_skills', hue = 'median', ax = ax[0], palette = 'dark:b_r')
ax[0].legend().remove()
ax[0].set_title('Top 10 Highest paid Skills for Data Analysts')
ax[0].set_ylabel('')
ax[0].set_xlabel('')
ax[0].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f'${int(x/1000)}K'))

# Top 10 Most In-Demand Skills for Data Analysts
sns.barplot(data = df_DA_top_skills, x = 'median', y = 'job_skills', hue = 'median', ax = ax[1], palette = 'light:b')
ax[1].legend().remove()
ax[1].set_title('Top 10 Most In-Demand Skills for Data Analysts')
ax[1].set_ylabel('')
ax[1].set_xlabel('')
ax[1].set_xlim(ax[0].get_xlim())
ax[1].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f'${int(x/1000)}K'))

plt.tight_layout()
```
